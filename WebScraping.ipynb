#Importar as Bibliotecas

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

#Vamos da uma olha na página que vamos fazer scraping

##Vamos inspecionar ela

https://pt.accio.com/business/idiomas-mais-populares-no-mundo

##Depois vamos coletar a página

```python
html=requests.get("https://pt.accio.com/business/idiomas-mais-populares-no-mundo").content
soup=BeautifulSoup(html, 'html5lib')
```

##Com a biblioteca Beatifulsoup vamos encontar o primeiro parágrado, pegar a tag P

```python
primeiro_paragrafo=soup.find('p')
primeiro_paragrafo
```

```python
primeiro_paragrafo.text
```

```python
todos_paragrafos=soup.find_all('p')
todos_paragrafos
```

```python
todos_paragrafos[0]
```

##Agora vamos pegar todos os links

```python
todoslinks=soup.find_all('a')
todoslinks
```

##Vamos inspecionar a tabela, para pegar os dados da tabela "Principais idiomas falados"

```python
tabela=soup.find('table').find('tbody')
tabela
```

```python
linhas=tabela.find_all('tr')
for linha in linhas:
  dado=linha.find_all('td')
  print(dado[0].text)
  print(dado[1].text)
  print(dado[2].text)
  print('-------')
```

##Mostrar uma lista de idiomas e de porcentagem com os dados, para depois compor um dataframe

```python
idiomas=[]
pontos=[]
for linha in linhas:
  dado=linha.find_all('td')
  idiomas.append(dado[0].text)
  pontos.append(dado[1].text)
```

```python
print(idiomas)
print(pontos)
```

```python
dados=pd.DataFrame(idiomas, columns=['Idiomas'])
```

```python
dados['Pontos']=pontos
```
